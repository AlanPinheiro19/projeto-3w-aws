# Dashboard 3W - Analise de Features e Classes
## TCC: Sistema para Deteccao e Classificacao de Eventos em Pocos Offshore

Dashboard interativo com Plotly. Execute todas as celulas (Run All) para carregar os graficos.

```python
import os
import sys
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

warnings.filterwarnings('ignore')

# Instala plotly se necessario
try:
    import plotly
    import plotly.graph_objects as go
    import plotly.express as px
    from plotly.subplots import make_subplots
    print('Plotly OK:', plotly.__version__)
except ImportError:
    import subprocess
    subprocess.run([sys.executable, '-m', 'pip', 'install', 'plotly', '-q'], check=True)
    import plotly
    import plotly.graph_objects as go
    import plotly.express as px
    from plotly.subplots import make_subplots
    print('Plotly instalado e OK:', plotly.__version__)

PROJECT_ROOT = Path(os.environ.get('PROJECT_DIR', '/home/jovyan/workspace'))

SENSOR_COLUMNS = ['P-PDG', 'P-TPT', 'T-TPT', 'P-MON-CKP', 'T-JUS-CKP', 'P-JUS-CKGL', 'T-JUS-CKGL', 'QGL']

EVENT_LABELS = {
    0: 'Normal',
    1: 'BSW Abrupt Increase',
    2: 'DHSV Spurious Closure',
    3: 'Severe Slugging',
    4: 'Flow Instability',
    5: 'Productivity Loss',
    6: 'Restriction in PCK',
    7: 'Scaling in PCK',
    8: 'Hydrate in Pipeline',
    9: 'Undefined',
}

COLOR_MAP = px.colors.qualitative.T10
print('Setup OK | PROJECT_ROOT:', PROJECT_ROOT)
```

```python
def load_parquets(path: Path, pattern='*.parquet') -> pd.DataFrame:
    files = list(path.glob(pattern))
    if not files:
        print(f'[AVISO] Sem dados em: {path}')
        return pd.DataFrame()
    df = pd.concat([pd.read_parquet(f) for f in files], ignore_index=True)
    print(f'{path.name}: {len(df):,} linhas | {df.shape[1]} colunas')
    return df

df_silver = load_parquets(PROJECT_ROOT / 'data' / 'silver')
df_train  = load_parquets(PROJECT_ROOT / 'data' / 'gold', 'train.parquet')
df_val    = load_parquets(PROJECT_ROOT / 'data' / 'gold', 'val.parquet')
df_test   = load_parquets(PROJECT_ROOT / 'data' / 'gold', 'test.parquet')

if not df_silver.empty and 'class' in df_silver.columns:
    df_silver['class'] = df_silver['class'].astype(int)
    df_silver['classe_label'] = df_silver['class'].map(EVENT_LABELS)

print('\nDados carregados com sucesso.')
```

---
## 1. Distribuicao de Classes

```python
if not df_silver.empty and 'class' in df_silver.columns:
    counts = df_silver['class'].value_counts().sort_index().reset_index()
    counts.columns = ['class', 'count']
    counts['label'] = counts['class'].map(EVENT_LABELS)
    counts['pct'] = (counts['count'] / counts['count'].sum() * 100).round(2)
    counts['text'] = counts['pct'].map(lambda x: f'{x:.1f}%')
    counts['color'] = [COLOR_MAP[i % len(COLOR_MAP)] for i in range(len(counts))]

    fig = make_subplots(
        rows=1, cols=2,
        subplot_titles=('Contagem por Classe', 'Proporcao (%)'),
        specs=[[{'type': 'bar'}, {'type': 'pie'}]]
    )

    fig.add_trace(go.Bar(
        x=counts['label'],
        y=counts['count'],
        text=counts['text'],
        textposition='outside',
        marker_color=counts['color'],
        hovertemplate='<b>%{x}</b><br>Count: %{y:,}<br>%{text}<extra></extra>',
        name=''
    ), row=1, col=1)

    fig.add_trace(go.Pie(
        labels=counts['label'],
        values=counts['count'],
        hole=0.4,
        marker_colors=counts['color'],
        hovertemplate='<b>%{label}</b><br>%{value:,} registros<br>%{percent}<extra></extra>',
        textinfo='percent+label'
    ), row=1, col=2)

    fig.update_layout(
        title='Distribuicao de Classes de Evento (Silver)',
        height=420,
        showlegend=False,
        xaxis_tickangle=-35
    )
    fig.show()

    total = counts['count'].sum()
    ratio = counts['count'].max() / counts['count'].min()
    print(f'Total: {total:,} registros | {len(counts)} classes | Imbalance ratio: {ratio:.1f}x')
```

---
## 2. Distribuicao dos Sensores por Classe (Violin + Box)

```python
if not df_silver.empty:
    sensor_cols = [c for c in SENSOR_COLUMNS if c in df_silver.columns]

    # Widget de selecao de sensor
    from ipywidgets import interact, Dropdown

    def plot_sensor(sensor):
        sample = df_silver[['classe_label', sensor]].dropna()
        if len(sample) > 20000:
            sample = sample.sample(20000, random_state=42)

        classes_ordered = sorted(df_silver['class'].unique())
        labels_ordered = [EVENT_LABELS[c] for c in classes_ordered]

        fig = go.Figure()
        for i, (cls, lbl) in enumerate(zip(classes_ordered, labels_ordered)):
            vals = sample.loc[sample['classe_label'] == lbl, sensor]
            if len(vals) < 5:
                continue
            fig.add_trace(go.Violin(
                y=vals,
                name=f'C{cls}: {lbl}',
                box_visible=True,
                meanline_visible=True,
                fillcolor=COLOR_MAP[i % len(COLOR_MAP)],
                opacity=0.7,
                line_color='black',
                line_width=0.8,
            ))

        fig.update_layout(
            title=f'Distribuicao de {sensor} por Classe',
            yaxis_title=sensor,
            height=450,
            showlegend=True,
            violingap=0.1,
            violinmode='overlay'
        )
        fig.show()

    interact(plot_sensor, sensor=Dropdown(options=sensor_cols, description='Sensor:'))
```

---
## 3. Heatmap de Correlacao dos Sensores

```python
if not df_silver.empty:
    sensor_cols = [c for c in SENSOR_COLUMNS if c in df_silver.columns]
    sample = df_silver[sensor_cols].sample(min(10000, len(df_silver)), random_state=42)
    corr = sample.corr().round(3)

    fig = go.Figure(go.Heatmap(
        z=corr.values,
        x=corr.columns,
        y=corr.index,
        colorscale='RdBu',
        zmid=0,
        text=corr.values.round(2),
        texttemplate='%{text}',
        hovertemplate='%{y} x %{x}<br>Correlacao: %{z:.3f}<extra></extra>'
    ))
    fig.update_layout(
        title='Matriz de Correlacao dos Sensores (Silver)',
        height=500,
        width=600
    )
    fig.show()

    print('Pares com |correlacao| > 0.7:')
    for i in range(len(corr)):
        for j in range(i+1, len(corr)):
            v = corr.iloc[i, j]
            if abs(v) > 0.7:
                print(f'  {corr.columns[i]} <-> {corr.index[j]}: {v:.3f}')
```

---
## 4. Serie Temporal Interativa por Poco

```python
if not df_silver.empty and 'timestamp' in df_silver.columns and 'ID_Poco' in df_silver.columns:
    from ipywidgets import interact, Dropdown

    df_silver['timestamp'] = pd.to_datetime(df_silver['timestamp'])
    sensor_cols = [c for c in SENSOR_COLUMNS if c in df_silver.columns]
    pocos = sorted(df_silver['ID_Poco'].unique())

    def plot_timeseries(poco, sensor):
        df_p = df_silver[df_silver['ID_Poco'] == poco].sort_values('timestamp')
        if df_p.empty:
            print('Sem dados para este poco.')
            return

        # Limita a 50k pontos para performance
        if len(df_p) > 50000:
            step = len(df_p) // 50000 + 1
            df_p = df_p.iloc[::step]

        fig = go.Figure()
        classes_present = sorted(df_p['class'].unique())
        for cls in classes_present:
            mask = df_p['class'] == cls
            fig.add_trace(go.Scattergl(
                x=df_p.loc[mask, 'timestamp'],
                y=df_p.loc[mask, sensor],
                mode='markers',
                marker=dict(size=2, color=COLOR_MAP[cls % len(COLOR_MAP)], opacity=0.6),
                name=f'C{cls}: {EVENT_LABELS.get(cls, "?")}',
                hovertemplate=f'%{{x}}<br>{sensor}: %{{y:.2f}}<br>Classe {cls}<extra></extra>'
            ))

        fig.update_layout(
            title=f'Serie Temporal - Poco {poco} | {sensor}',
            xaxis_title='Timestamp',
            yaxis_title=sensor,
            height=400,
            hovermode='x unified',
            xaxis=dict(rangeslider=dict(visible=True), type='date')
        )
        fig.show()

    interact(
        plot_timeseries,
        poco=Dropdown(options=pocos, description='Poco:'),
        sensor=Dropdown(options=sensor_cols, description='Sensor:')
    )
```

---
## 5. Estatisticas Descritivas por Classe e Sensor

```python
if not df_silver.empty:
    from ipywidgets import interact, SelectMultiple, Dropdown
    sensor_cols = [c for c in SENSOR_COLUMNS if c in df_silver.columns]

    def plot_stats(metric, sensors):
        if not sensors:
            print('Selecione ao menos um sensor.')
            return

        agg = df_silver.groupby('classe_label')[list(sensors)].agg(metric)
        agg = agg.reset_index().melt(id_vars='classe_label', var_name='sensor', value_name='valor')

        fig = px.bar(
            agg,
            x='sensor',
            y='valor',
            color='classe_label',
            barmode='group',
            title=f'{metric.upper()} dos Sensores por Classe',
            labels={'valor': metric, 'sensor': 'Sensor', 'classe_label': 'Classe'},
            color_discrete_sequence=COLOR_MAP,
            height=420
        )
        fig.update_layout(xaxis_tickangle=-30)
        fig.show()

    interact(
        plot_stats,
        metric=Dropdown(
            options=['mean', 'std', 'median', 'min', 'max'],
            value='mean',
            description='Metrica:'
        ),
        sensors=SelectMultiple(
            options=sensor_cols,
            value=sensor_cols[:4],
            description='Sensores:',
            rows=len(sensor_cols)
        )
    )
```

---
## 6. Analise das Features Engineered (Camada Gold)

```python
if not df_train.empty:
    norm_cols = [c for c in df_train.columns if c.endswith('_norm')]
    total = len(df_train) + len(df_val) + len(df_test)

    # Painel de splits
    fig = make_subplots(
        rows=1, cols=3,
        subplot_titles=(
            f'Treino ({len(df_train):,})',
            f'Validacao ({len(df_val):,})',
            f'Teste ({len(df_test):,})'
        )
    )

    for col_idx, (name, df_split) in enumerate(
        [('Treino', df_train), ('Validacao', df_val), ('Teste', df_test)], start=1
    ):
        if df_split.empty or 'class' not in df_split.columns:
            continue
        counts = df_split['class'].astype(int).value_counts().sort_index()
        fig.add_trace(go.Bar(
            x=[f'C{k}' for k in counts.index],
            y=counts.values,
            marker_color=[COLOR_MAP[k % len(COLOR_MAP)] for k in counts.index],
            showlegend=False,
            hovertemplate='Classe %{x}<br>%{y:,} registros<extra></extra>'
        ), row=1, col=col_idx)

    fig.update_layout(
        title=f'Distribuicao de Classes por Split (Gold) | Total: {total:,}',
        height=380
    )
    fig.show()
```

```python
if not df_train.empty:
    from ipywidgets import interact, Dropdown

    norm_cols = sorted([c for c in df_train.columns if c.endswith('_norm')])

    def plot_feature_dist(feature):
        fig = go.Figure()

        for split_name, df_split, color in [
            ('Treino', df_train, 'royalblue'),
            ('Validacao', df_val, 'orange'),
            ('Teste', df_test, 'green')
        ]:
            if df_split.empty or feature not in df_split.columns:
                continue
            vals = df_split[feature].dropna()
            fig.add_trace(go.Histogram(
                x=vals,
                name=split_name,
                opacity=0.6,
                nbinsx=60,
                marker_color=color,
                hovertemplate=f'{split_name}<br>Z-score: %{{x:.2f}}<br>Count: %{{y}}<extra></extra>'
            ))

        sensor_base = feature.replace('_norm', '').replace('_', '-').upper()
        fig.update_layout(
            title=f'Distribuicao Z-score: {sensor_base}',
            xaxis_title='Z-score',
            yaxis_title='Frequencia',
            barmode='overlay',
            height=380
        )
        fig.show()

    interact(
        plot_feature_dist,
        feature=Dropdown(options=norm_cols, description='Feature:')
    )
```

---
## 7. Variancia das Features por Classe (Importancia Proxy)

```python
if not df_silver.empty:
    sensor_cols = [c for c in SENSOR_COLUMNS if c in df_silver.columns]

    # Variancia entre-classes (desvio padrao das medias por classe)
    class_means = df_silver.groupby('class')[sensor_cols].mean()
    between_class_var = class_means.std().sort_values(ascending=True)

    fig = go.Figure(go.Bar(
        x=between_class_var.values,
        y=between_class_var.index,
        orientation='h',
        marker_color=[
            f'rgba(220,50,50,0.8)' if v > between_class_var.median()
            else 'rgba(50,100,200,0.6)'
            for v in between_class_var.values
        ],
        hovertemplate='%{y}<br>Variancia entre classes: %{x:.4f}<extra></extra>'
    ))

    fig.update_layout(
        title='Poder Discriminativo dos Sensores (desvio padrao das medias por classe)',
        xaxis_title='Desvio padrao das medias por classe',
        yaxis_title='Sensor',
        height=380
    )
    fig.show()

    best = between_class_var.idxmax()
    worst = between_class_var.idxmin()
    print(f'Sensor com maior poder discriminativo: {best} ({between_class_var[best]:.4f})')
    print(f'Sensor com menor poder discriminativo: {worst} ({between_class_var[worst]:.4f})')
```

---
## 8. Nulos por Sensor e Camada

```python
if not df_silver.empty:
    sensor_cols = [c for c in SENSOR_COLUMNS if c in df_silver.columns]
    null_pct = (df_silver[sensor_cols].isnull().mean() * 100).round(2)

    fig = go.Figure(go.Bar(
        x=null_pct.index,
        y=null_pct.values,
        text=[f'{v:.1f}%' for v in null_pct.values],
        textposition='outside',
        marker_color=[
            'rgba(220,50,50,0.8)' if v > 5
            else 'rgba(50,180,100,0.7)'
            for v in null_pct.values
        ],
        hovertemplate='%{x}<br>Nulos: %{y:.2f}%<extra></extra>'
    ))

    fig.add_hline(y=5, line_dash='dash', line_color='red',
                  annotation_text='Limite 5%', annotation_position='right')

    fig.update_layout(
        title='Percentual de Valores Nulos por Sensor (Silver)',
        xaxis_title='Sensor',
        yaxis_title='% Nulos',
        yaxis_range=[0, max(null_pct.max() * 1.3, 10)],
        height=380
    )
    fig.show()
```

---
## 9. Resumo Executivo

```python
print('=' * 65)
print(' RESUMO EXECUTIVO - Dashboard 3W')
print('=' * 65)

if not df_silver.empty:
    counts = df_silver['class'].value_counts()
    sensor_cols = [c for c in SENSOR_COLUMNS if c in df_silver.columns]
    null_max = (df_silver[sensor_cols].isnull().mean() * 100).max()
    class_means = df_silver.groupby('class')[sensor_cols].mean()
    best_sensor = class_means.std().idxmax()

    print(f'\n[Dados]')
    print(f'  Registros Silver      : {len(df_silver):,}')
    print(f'  Classes presentes     : {sorted(df_silver["class"].unique().tolist())}')
    print(f'  Imbalance ratio       : {counts.max()/counts.min():.1f}x')
    print(f'  Classe majoritaria    : {int(counts.idxmax())} - {EVENT_LABELS.get(int(counts.idxmax()),"?")} ({counts.max():,})')
    print(f'  Nulos max (Silver)    : {null_max:.1f}%')
    print(f'  Sensor discriminativo : {best_sensor}')

if not df_train.empty:
    norm_cols = [c for c in df_train.columns if c.endswith('_norm')]
    total = len(df_train) + len(df_val) + len(df_test)
    print(f'\n[Camada Gold - ML Ready]')
    print(f'  Total registros       : {total:,}')
    print(f'  Features normalizadas : {len(norm_cols)}')
    print(f'  Treino / Val / Teste  : {len(df_train):,} / {len(df_val):,} / {len(df_test):,}')

print(f'\n[Proximas Etapas]')
print(f'  - Treinar Random Forest (baseline)')
print(f'  - Treinar XGBoost e LightGBM')
print(f'  - Comparar F1-weighted entre modelos')
print(f'  - Avaliar com matriz de confusao')
print('=' * 65)
```
